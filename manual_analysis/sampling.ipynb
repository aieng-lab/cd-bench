````python
# load json to pandas df

import os
import pandas as pd

# check if mutants_diffs_sampled.json exists


if not os.path.exists('mutants_diffs_sampled.json'):
    print('mutants_diffs_sampled.json does not exist, creating it now.')
    # create sampled json
    n_samples = 50
    df = pd.read_json('mutants_diffs.json', lines=True)

    # sample n_samples per attacker
    df_sampled = df.groupby('attacker').apply(lambda x: x.sample(n=min(len(x), n_samples), random_state=1)).reset_index(drop=True)
    # shuffle rows
    df_sampled = df_sampled.sample(frac=1, random_state=1).reset_index(drop=True)
    # store sampled df to json
    df_sampled.to_json('mutants_diffs_sampled.json', orient='records', lines=True)
else:
    print('mutants_diffs_sampled.json already exists, loading it directly.')
    df_sampled = pd.read_json('mutants_diffs_sampled.json', lines=True)

````

````python
import ipywidgets as widgets
from IPython.display import display, Markdown

def diff_for_annotation(df_sampled, index):
    mutant_id = df_sampled.iloc[index]['mutant_id']
    equivalent = df_sampled.iloc[index]['equivalent']
    mutant = df_sampled.iloc[index]['diff']

    display(Markdown(f'### Mutant ID: {mutant_id}'))
    display(Markdown(f'**Equivalent:** {equivalent}'))
    display(Markdown(f'```diff\n{mutant}\n```'))

    text_annot = widgets.Text(
        value='',
        description='Annotation:',
        disabled=False
    )
    display(text_annot)

for i in range(20):
    print(i)
    diff_for_annotation(df_sampled, i)
````
