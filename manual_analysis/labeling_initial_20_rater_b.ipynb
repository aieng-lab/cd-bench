# Annotations for initial 20% of sampled mutants by Rater B

Guidelines for annotations:
- The form shows whether our test suite determined equivalence or not and the diff of the mutant
- Lines that have the comment "// This line is already mutated, changing it again is forbidden" were not mutated. This is part of the diff for technical reasons, as the LLM got this as part of the instruction *not* to mutate these lines. Please ignore these lines when making your decision.
- The form is simple. You have to annotate everything in a single pass and then store by running the last cell. Storing partial results is not supported. 

````python
# Run this cell to create annotation form

import math
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, Markdown

def diff_for_annotation(df_sampled, index):
    mutant_id = df_sampled.iloc[index]['mutant_id']
    equivalent = df_sampled.iloc[index]['equivalent']
    mutant = df_sampled.iloc[index]['diff']

    display(Markdown(f'### Mutant ID: {mutant_id}'))
    display(Markdown(f'**Equivalent:** {equivalent}'))
    display(Markdown(f'```diff\n{mutant}\n```'))

    text_annot = widgets.Text(
        value='',
        description='Annotation:',
        disabled=False
    )
    display(text_annot)
    return mutant_id, equivalent, text_annot

df_sampled = pd.read_json('data/mutants_diffs_sampled.json', lines=True)

start_index = 0
end_index = math.ceil(len(df_sampled)*0.2)

print(f"Displaying mutants from index {start_index} to {end_index - 1}")

labeling_results = {}
for i in range(start_index, end_index):
    mutant_id, equivalent, text_annot = diff_for_annotation(df_sampled, i)
    labeling_results[mutant_id] = {
        'equivalent': equivalent,
        'annotation': text_annot
    }
````

````python
# run this cell to save results

coded_data = pd.DataFrame.from_dict(labeling_results, orient='index')
coded_data['annotation'] = coded_data['annotation'].apply(lambda x: x.value)

coded_data.to_csv('data/labeling_initial_20_rater_b.csv')
````
