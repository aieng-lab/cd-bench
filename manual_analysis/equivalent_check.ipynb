```python
import pandas as pd
import ipywidgets as widgets
import ghdiff
from IPython.display import display, Markdown

def diff_for_annotation(df_sampled, index):
    mutant_id = df_sampled.iloc[index]['mutant_id']
    class_alias = df_sampled.iloc[index]['class_alias']
    mutant = df_sampled.iloc[index]['diff']

    display(Markdown(f'### Mutant ID: {mutant_id} - Index: {index} - Class: {class_alias}'))
    diff = widgets.HTML(
        value=ghdiff.colorize(mutant),
        disabled=True,
        layout=widgets.Layout(flex='1 1 auto', width='50% width', height='auto')
    )

    radio_labels = widgets.RadioButtons(
        options=['equivalent', 'non-equivalent', 'uncertain'],
        description='Label:',
        disabled=False,
        layout=widgets.Layout(flex='2 1 auto', width='auto', height='auto')
    )
    

    text_annot = widgets.Text(
        value='',
        disabled=False,
        layout=widgets.Layout(flex='2 2 auto', width='auto', height='auto')
    )

    annot_area = widgets.VBox(children=[radio_labels, text_annot],
                             layout=widgets.Layout(display='flex',
                                                   flex_flow='column',
                                                   align_items='stretch',
                                                   width='auto',
                                                   height='auto'))

    autolayout_box = widgets.Box(children=[diff, annot_area],
                                 layout=widgets.Layout(display='flex', width='80%'))

    display(autolayout_box)
    return mutant_id, radio_labels, text_annot

df_sampled = pd.read_json('data/mutants_diffs_sampled.json', lines=True)
df_sampled = df_sampled[df_sampled['equivalent']==True]
```

```python
start_index = 0
end_index = len(df_sampled)

print(f"Displaying mutants from index {start_index} to {end_index - 1}")

labeling_results = {}
for i in range(start_index, end_index):
    mutant_id, radio_labels, text_annot = diff_for_annotation(df_sampled, i)
    labeling_results[mutant_id] = {
        'label': radio_labels,
        'annotation': text_annot
    }
```
