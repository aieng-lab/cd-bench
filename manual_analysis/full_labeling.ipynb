# Labeling Notebook

This is the labeling notebook for the full data. The predefined labels are the outcome of the labeling of the first 20% of the data by two raters. New codes are added as needed. For this, select other and use the free text field to define the code.

To conduct the labeling, run the cells in order below. The first cell defines the required functions, the next three cells load 100 mutants each for labeling. The last cell saves the labeled data to a CSV file.

DO NOT USE RUN ALL CELLS! This might overwrite data. Make sure to only uncomment the last cell when you want to save data. 

Note that widgets typically do not render correctly after reloading the notebook, so the labeling needs to be done in a single session. 

```python
import pandas as pd
import ipywidgets as widgets
import ghdiff
from IPython.display import display, Markdown

predefined_labels = [
    'Delete conditional behavior',
    'Delete statement (assignment)',
    'Delete statement (control flow)',
    'Delete statement (method call)',
    'Delete terms from arithmetic expression',
    'Empty diff',
    'High-order mutant: Replace conditional operator (negation), Insert conditional behavior',
    'High-order mutant: Replace pre-increment with post-increment, Replace conditional operator (boundary)',
    'Insert bit shift',
    'Insert concatenation with empty string',
    'Insert conditional behaviour',
    'Insert literal (numeric)',
    'Insert masking',
    'Insert method call',
    'Insert term into arithmetic operation',
    'Insert unary operator (numeric negation)',
    'Replace == with equals',
    'Replace arithmetic operator',
    'Replace arithmetic operator (2nd order)',
    'Replace bit shift operator',
    'Replace conditional operator (boundary)',
    'Replace conditional operator (negation)',
    'Replace literal value (boolean)',
    'Replace method call with literal',
    'Replace method call with other method',
    'Replace variable',
    'Replace variable (2nd order)',
    'Other'
]

def diff_for_annotation(df_sampled, index):
    mutant_id = df_sampled.iloc[index]['mutant_id']
    equivalent = df_sampled.iloc[index]['equivalent']
    mutant = df_sampled.iloc[index]['diff']

    display(Markdown(f'### Mutant ID: {mutant_id} - Index: {index}'))
    display(Markdown(f'**Equivalent:** {equivalent}'))
    diff = widgets.HTML(
        value=ghdiff.colorize(mutant),
        disabled=True,
        layout=widgets.Layout(flex='1 1 auto', width='50% width', height='auto')
    )

    radio_labels = widgets.RadioButtons(
        options=predefined_labels,
        description='Label:',
        disabled=False,
        layout=widgets.Layout(flex='2 1 auto', width='auto', height='auto')
    )
    

    text_annot = widgets.Text(
        value='',
        disabled=False,
        layout=widgets.Layout(flex='2 2 auto', width='auto', height='auto')
    )

    annot_area = widgets.VBox(children=[radio_labels, text_annot],
                             layout=widgets.Layout(display='flex',
                                                   flex_flow='column',
                                                   align_items='stretch',
                                                   width='auto',
                                                   height='auto'))

    autolayout_box = widgets.Box(children=[diff, annot_area],
                                 layout=widgets.Layout(display='flex', width='80%'))

    display(autolayout_box)
    return mutant_id, equivalent, radio_labels, text_annot

df_sampled = pd.read_json('data/mutants_diffs_sampled.json', lines=True)
```

```python
start_index = 0
end_index = 10

print(f"Displaying mutants from index {start_index} to {end_index - 1}")

labeling_results = {}
for i in range(start_index, end_index):
    mutant_id, equivalent, radio_labels, text_annot = diff_for_annotation(df_sampled, i)
    labeling_results[mutant_id] = {
        'equivalent': equivalent,
        'label': radio_labels,
        'annotation': text_annot
    }
```
