```python
import pandas as pd
import numpy as np

df = pd.read_csv('data/mutants.csv')
```

```python
metrics = []
for (class_alias, attacker, defender), data in df.groupby(['class_alias', 'attacker', 'defender']):
    m_success = data[data['forfeit'] == False]
    m_noneq = data[data['equivalent'] == False]

    score = m_noneq[m_noneq['surviving'] == True].shape[0] / data.shape[0]

    mut_forfeit = 1 - m_success.shape[0] / data.shape[0]
    mut_equivalent = m_success[m_success['equivalent'] == True].shape[0] / m_success.shape[0] if m_success.shape[0] > 0 else np.nan
    mut_survived_turn = m_noneq[m_noneq['killed_immediately']
                                == False].shape[0] / m_noneq.shape[0] if m_noneq.shape[0] > 0 else np.nan
    mut_survived_game = m_noneq[m_noneq['surviving'] == True].shape[0] / m_noneq.shape[0] if m_noneq.shape[0] > 0 else np.nan

    metrics.append({
        'class_alias': class_alias,
        'attacker': attacker,
        'defender': defender,
        'score': score,
        'mut_forfeit': mut_forfeit,
        'mut_equivalent': mut_equivalent,
        'mut_survived_turn': mut_survived_turn,
        'mut_survived_game': mut_survived_game
    })

df_metrics = pd.DataFrame(metrics)
df_metrics.to_csv('data/mutant_metrics.csv', index=False)
```
