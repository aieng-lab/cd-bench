```python
import pandas as pd
import numpy as np

df = pd.read_csv('data/mutants.csv')
df_equivalents = pd.read_csv('data/equivalents.csv')
df_enriched = pd.read_csv('data/equivalents remapping.csv')
```

```python
df_equivalents['equivalent'] = df_equivalents['KillingTests'].isna()

df_equivalents = df_equivalents[['Classname', 'SetupID', 'MutantID', 'equivalent']]
df_merged = df_enriched.merge(df_equivalents, how='left', left_on=[
                              'Classname', 'SetupID', 'MutantID'], right_on=['Classname', 'SetupID', 'MutantID'])
df_main_join = df.merge(df_merged, how='left', left_on=['class_alias', 'turn', 'game_id'], right_on=[
    'Classname', 'turn', 'game_id'])
```

```python
df_main_join.to_csv('data/mutants.csv', index=False)
```
