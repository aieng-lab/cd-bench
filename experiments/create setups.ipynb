Creation of the setups to utilize further in the tournament.

```python
classes = [
    'Complex',
    'Rational',
    'Option',
    'XmlElement',
    'SparseIntArray',
    'IntHashMap',
    'ByteVector',
    'CharRange',
    'CaseInsensitiveString',
    'Document'
]


llms = [
    ('claude-3-5-haiku-latest', 'anthropic'),
    ('claude-sonnet-4-0', 'anthropic'),
    ('gpt-4.1-mini', 'openai'),
    ('gpt-4.1', 'openai'),
    ('gemini-2.5-flash', 'google'),
    ('gemini-2.5-pro', 'google')
    ]
```

```python
from util.models.battleground.battleground_game import BattlegroundGameSettings
import json

setup_id = 0
setups_path = 'resources/setups/'

for class_alias in classes:
    for attacker_llm_name, attacker_llm_family in llms:
        for defender_llm_name, defender_llm_family in llms:
            game_settings = BattlegroundGameSettings(
                class_alias,
                mutant_validator_level='relaxed',
                duration_minutes=60*24,
                level='easy',
                turns=20,
                attacker_mutants_per_turn=1,
                attacker_max_stillborn_mutants=3,
                attacker_max_compile_attempts=5,
                defender_max_kill_attempts=3,
                defender_max_compile_attempts=5,
                attacker_model_name=attacker_llm_name,
                attacker_model_family=attacker_llm_family,
                defender_model_name=defender_llm_name,
                defender_model_family=defender_llm_family
            )

            with open(f'{setups_path}{setup_id}.json', 'w') as file:
                json.dump(game_settings.to_json(), file, indent=4)

            setup_id += 1
```
